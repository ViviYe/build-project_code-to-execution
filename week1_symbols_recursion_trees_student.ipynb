# Part 3: Recursion

A recursive function is a function that calls itself.

Recursion is useful when a problem contains a smaller version of the same problem.

Classic example: countdown.

````python
def countdown(n):
    if n == 0:
        print("Done!")
    else:
        print(n)
        countdown(n - 1)

countdown(5)
````

Every recursive function needs:

1. A **base case**: when to stop
2. A **recursive case**: how to make the problem smaller

# Part 4: Trees

A tree is a structure made of nodes.

Each node can have children.

Example:


This tree represents:

```text

```

The root is `+`, and its children are `2` and `3`.

We can represent a tree in Python using dictionaries.

A number node might look like:


# Part 5: Tree traversal

Tree traversal means visiting each node in a tree.

For our interpreter, the most important question is:

> How do we evaluate this tree?

For a number node, evaluation is easy: return the number.

For a binary operator node, we:
1. Evaluate the left side
2. Evaluate the right side
3. Apply the operator

````python
def evaluate(node):
    if node["type"] == "number":
        return node["value"]

    elif node["type"] == "binary":
        left_value = evaluate(node["left"])
        right_value = evaluate(node["right"])
        operator = node["operator"]

        return apply_operator(left_value, operator, right_value)

    else:
        raise ValueError("Unknown node type: " + node["type"])

simple_tree = {
    "type": "binary",
    "operator": "+",
    "left": {"type": "number", "value": 2},
    "right": {"type": "number", "value": 3}
}

print(evaluate(simple_tree))  # expected: 5
````

# Part 7: Pretty-printing a tree

Interpreters often need to inspect internal structures.

Let’s write a function that prints the tree in a readable way.

````python
def print_tree(node, indent=0):
    spaces = " " * indent

    if node["type"] == "number":
        print(spaces + "Number(" + str(node["value"]) + ")")

    elif node["type"] == "binary":
        print(spaces + "Binary(" + node["operator"] + ")")
        print_tree(node["left"], indent + 2)
        print_tree(node["right"], indent + 2)

    else:
        print(spaces + "Unknown node")

print_tree(nested_tree)
````
