```python
def tokenize(text):
    tokens = []

    current = ""

    for ch in text:
        if ch.isdigit():
            current += ch
        else:
            if current:
                tokens.append(("NUMBER", current))
                current = ""
            tokens.append(("OPERATOR", ch))
            
    if current:
        tokens.append(("NUMBER", current))


    return tokens
```

```python
tokenize("12+2*100")
```

```python
class Number:
    def __init__(self, value):
        self.value = value
    def __repr__(self):
       return f"Number({self.value})"


class BinaryOp:
    def __init__(self, op, left, right):
        self.op = op
        self.left = left
        self.right = right
    def __repr__(self):
        return (
           f"BinaryOp({self.op}, "
           f"{self.left}, "
           f"{self.right})"
       )

```

```python
ast = BinaryOp(
    "+",
    Number(1),
    Number(2)
)

print(ast)
```

```python
tokens = [
   ("NUMBER", 1),
   ("PLUS", "+"),
   ("NUMBER", 2),
]
```

```python
def parse_simple_expression(tokens):
   left = Number(tokens[0][1])
   right = Number(tokens[2][1])

   op = tokens[1][1]

   return BinaryOp(op, left, right)
```
