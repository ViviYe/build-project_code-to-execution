# Part 1 — Quick Recursion Recap

A recursive function usually contains two important parts:

1. **Base case:** the condition that stops the recursion.
2. **Recursive case:** the step that reduces the problem and calls the function again.

```python
def countdown(n):
    if n <= 0:          # Base case
        return
    print(n)
    countdown(n - 1)    # Recursive case
```

# Part 2 — Visualizing Calls and Returns

A recursive function creates a chain of calls, then returns through that chain in reverse order.

````python
def traced_factorial(n, depth=0):
    indent = "  " * depth
    print(f"{indent}CALL factorial({n})")

    if n <= 1:
        print(f"{indent}BASE CASE -> return 1")
        return 1

    smaller_result = traced_factorial(n - 1, depth + 1)
    result = n * smaller_result

    print(f"{indent}RETURN factorial({n}) = {n} * {smaller_result} = {result}")
    return result

answer = traced_factorial(4)
print("\nFinal answer:", answer)
````

## Environment-frame model

Conceptually, recursive execution looks like this:

```text
Global Environment
    ↓
Call Environment: n = 4
    ↓
Call Environment: n = 3
    ↓
Call Environment: n = 2
    ↓
Call Environment: n = 1
```

Each call must have an independent local environment.

````python
def show_environment_chain(n, depth=0):
    indent = "  " * depth
    environment = {"n": n}

    print(f"{indent}Create environment {depth}: {environment}")

    if n <= 1:
        print(f"{indent}Base case in environment {depth}")
        return 1

    result = n * show_environment_chain(n - 1, depth + 1)
    print(f"{indent}Destroy environment {depth}; return {result}")
    return result

show_environment_chain(4)
````

# Part 6 — Why Recursion Should Work in Your Interpreter

Your interpreter should not need a special `RecursiveCallNode`.

A recursive call uses the same machinery as every other call:

```text
1. Find the function.
2. Evaluate arguments.
3. Create a fresh local environment.
4. Bind parameters.
5. Link to the correct parent environment.
6. Evaluate the body.
7. Return the result.
```

# Part 8 — Diagnostic Guide

| Symptom | Likely cause |
|---|---|
| Function name is undefined inside its own body | Function lookup cannot reach the defining environment |
| Every call sees the same parameter value | Local environment is reused |
| Base case runs, but final result is missing | Return values are not propagated |
| Nested calls work, but recursive calls do not | Function is not bound before its body is evaluated |
| Program never stops | Base case or recursive step is wrong |
| Correct for one call, wrong after a second | Mutable state or environment is shared |
| Parser rejects the program | Calls or returns are not parsed correctly |
| Stack overflow on small inputs | Argument is not moving toward the base case |
